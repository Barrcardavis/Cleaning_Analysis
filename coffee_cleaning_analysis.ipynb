```python
# Replace invalid values
exports_df = exports_df.replace(-2147483648, pd.NA)

# Verify
(exports_df == -2147483648).sum().sum()
```

Three export values in the Brazil record were identified as -2147483648. Since coffee export volumes cannot be negative and the value corresponds to the minimum 32-bit integer limit, these records were treated as invalid data and replaced with missing values (NA).

# Step 4.1 - Largest Production Change by Country

## Business Question 1

Which country experienced the largest production change over the available reporting period?


This analysis identifies the country with the greatest change in coffee production between the first and last reporting periods. Production values are aggregated by country and the difference between 1990/91 and 2019/20 is calculated.

```python
# Aggregate production by country
country_change = (
    production_df
    .groupby("Country")[["1990/91", "2019/20"]]
    .sum()
    .reset_index()
)

# Calculate production change
country_change["Production_Change"] = (
    country_change["2019/20"]
    - country_change["1990/91"]
)

# Sort by largest increase
country_change = country_change.sort_values(
    by="Production_Change",
    ascending=False
)

country_change.head(10)
```

## Findings for Business Question 1
The results identify the countries with the largest increases and decreases in coffee production over the reporting period. Countries at the top of the table experienced the greatest production growth, indicating an expanded role in global coffee production.

# Step 4.2 - Trade Balance

## Business Question 2

Trade balance is calculated as total exports minus total imports. Positive values indicate a net exporter, while negative values indicate a net importer.

```python
trade_balance = pd.merge(
    exports_df[["Country", "Total_export"]],
    imports_df[["Country", "Total_import"]],
    on="Country",
    how="outer"
).fillna(0)

trade_balance["Trade_Balance"] = (
    trade_balance["Total_export"]
    - trade_balance["Total_import"]
)

trade_balance
```

## Findings for Business Question 2
Countries with positive trade balances are net exporters of coffee, while countries with negative trade balances are net importers. The results demonstrate the differing roles countries play in the global coffee trade network.

# Step 4.3 - Global Production Ratio by Coffee type

## Business Question 3

This analysis calculates the percentage of total global coffee production contributed by each coffee type. The production totals for each coffee type are aggregated and then converted into percentages of overall coffee production.

```python
# Calculate total production for each coffee type
coffee_type_totals = (
    production_df
    .groupby("Coffee type")["Total_production"]
    .sum()
)

# Calculate production percentages
production_ratio = (
    coffee_type_totals
    / coffee_type_totals.sum()
) * 100

# Display results
production_ratio.sort_values(ascending=False)
```

## Findings for Question 3
The results show the relative contribution of each coffee type to global coffee production. The coffee type with the highest percentage represents the largest share of worldwide production, providing insight into the dominant varieties supplied to the global coffee market.

# Step 4.4 - Countries Exporting Coffee with Zero Production

## Business Question 4

This analysis identifies countries that report coffee exports despite having no recorded coffee production in the production dataset. These countries may function as trade hubs, re-export centers, or may be affected by differences in data collection and reporting methods.


```python
# Calculate total production by country
production_by_country = (
    production_df
    .groupby("Country")["Total_production"]
    .sum()
    .reset_index()
)

# Merge export and production data
export_production = pd.merge(
    exports_df[["Country", "Total_export"]],
    production_by_country,
    on="Country",
    how="left"
)

# Identify countries with exports but no production
zero_production_exporters = export_production[
    (export_production["Total_export"] > 0)
    &
    (
        export_production["Total_production"].isna()
        |
        (export_production["Total_production"] == 0)
    )
]

# Display results
zero_production_exporters
```

## Findings for Business Question 4
The identified countries report coffee exports despite having no production records in the dataset. This may occur when countries import coffee, process or distribute it, and then re-export it to other markets. Another possible explanation is differences in dataset coverage or reporting practices between the production and export datasets.

# Step 5.1 Top 5 Net Exporters vs Top 5 Net Importers

## Visualization 1

This visualization compares the trade balances of the top five net exporting countries and the top five net importing countries. Positive values indicate countries that export more coffee than they import, while negative values indicate countries that import more coffee than they export.

```python
import matplotlib.pyplot as plt

# Top 5 net exporters
top_exporters = trade_balance.nlargest(5, "Trade_Balance")

# Top 5 net importers
top_importers = trade_balance.nsmallest(5, "Trade_Balance")

# Combine for plotting
trade_plot = pd.concat([top_exporters, top_importers])

# Create chart
plt.figure(figsize=(12, 6))
plt.bar(
    trade_plot["Country"],
    trade_plot["Trade_Balance"],
    color=[
        "green" if x > 0 else "red"
        for x in trade_plot["Trade_Balance"]
    ],
    label="Trade Balance"
)

plt.title("Top 5 Net Exporters vs Top 5 Net Importers")
plt.xlabel("Country")
plt.ylabel("Trade Balance")
plt.xticks(rotation=45)
plt.legend()
plt.tight_layout()
plt.show()
```

The chart highlights the countries with the largest positive and negative coffee trade balances. Net exporters contribute more coffee to global markets than they consume, while net importers rely heavily on imported coffee to satisfy domestic demand.

# Step 5.2 Global Coffee Export Trend Over Time

Visualization 2

This visualization shows the global trend of coffee exports over time by summing annual exports across all exporting countries. The chart helps identify long-term growth, declines, and fluctuations in worldwide coffee trade activity.

```python
# Select yearly export columns
export_years = [
    str(year)
    for year in range(1990, 2020)
]

# Calculate global exports by year
global_exports = exports_df[export_years].sum()

# Create line chart
plt.figure(figsize=(12, 6))
plt.plot(
    export_years,
    global_exports,
    marker="o",
    label="Global Coffee Exports"
)

plt.title("Global Coffee Export Trend (1990-2019)")
plt.xlabel("Year")
plt.ylabel("Total Exports")
plt.xticks(rotation=45)
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()
```

The line chart illustrates how worldwide coffee exports changed across the reporting period. Trends and fluctuations may reflect changes in production levels, international demand, weather events, and global economic conditions affecting coffee-producing countries.

# Step 5.3 Global Market Share by Coffee Type

This visualization displays the percentage contribution of each coffee type to total global production. The chart provides insight into the dominance of different coffee varieties within the global coffee industry.

```python
# Calculate production by coffee type
coffee_type_totals = (
    production_df
    .groupby("Coffee type")["Total_production"]
    .sum()
)

# Create pie chart
plt.figure(figsize=(8, 8))
plt.pie(
    coffee_type_totals,
    labels=coffee_type_totals.index,
    autopct="%1.1f%%",
    startangle=90
)

plt.title("Global Market Share by Coffee Type")
plt.legend(
    coffee_type_totals.index,
    title="Coffee Type",
    loc="best"
)

plt.tight_layout()
plt.show()
```

The visualization shows the relative share of total global coffee production attributed to each coffee type. Larger segments indicate coffee varieties that play a more significant role in worldwide coffee production and supply chains.

# Step 6: Executive Summary

This project analyzed global coffee production, imports, and exports to better understand how coffee moves through the worldwide supply chain. Before the analysis began, the data was reviewed and cleaned to correct formatting issues and remove invalid values. Once the data was cleaned, several trends became clear.

Coffee production is heavily concentrated in a small number of countries. Brazil stood out as one of the largest coffee\-producing nations and showed significant growth over the reporting period. Because so much of the world's coffee supply comes from a limited number of countries, any disruption in those regions could affect the entire market. Weather events, crop disease, labor shortages, and economic challenges all have the potential to reduce supply and increase costs.

The trade analysis showed a clear difference between countries that produce coffee and countries that consume it. Many European countries, along with the United States and Japan, import large amounts of coffee because they produce little or none themselves. In contrast, major producing countries export coffee to buyers around the world. This relationship highlights the importance of stable international trade and reliable supplier networks.

The analysis also showed that global coffee production is dominated by a relatively small number of coffee varieties. Relying too heavily on a limited range of products can create additional risk if production problems occur. Diversifying sourcing strategies across multiple countries and coffee types can help reduce that risk.

Another interesting finding was that some countries reported coffee exports even though they had no recorded coffee production in the dataset. This likely occurs because coffee is imported into those countries, processed or packaged, and then exported again to other markets. These countries often serve as trading and distribution hubs within the global coffee supply chain.

Based on these findings, supply\-chain leaders should focus on diversifying suppliers, monitoring production trends in major coffee\-producing countries, and maintaining backup sourcing options. Taking these steps can help reduce risk, improve supply stability, and better prepare the organization for future disruptions in the global coffee market.
